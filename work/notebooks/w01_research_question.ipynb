## 1. My lane (or freestyle) and why

I am choosing the Refresh / Content Opportunity Scoring lane. The reason is that it connects directly to an actionable decision: which existing content pages are most worth refreshing or improving next. This lane is attractive because the starter dataset already contains observable signals such as recent traffic, position, CTR, engagement, recency, and trend direction, so it is realistic to build an evidence-based prioritization approach without pretending to know the full product policy.

This is a provisional lane, not a final claim. I will use it to test whether the available signals can help rank pages in a way that is useful for a human reviewer.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd
from pathlib import Path


def starter_csv_path():
    cwd = Path.cwd().resolve()
    print('DEBUG cwd:', cwd)

    # First try the repo-relative path from the current working directory
    checked = []
    for folder in [cwd] + list(cwd.parents):
        candidate = folder / 'data' / 'raw' / 'content_refresh_anonymized.csv'
        checked.append(candidate)
        if candidate.exists():
            print('DEBUG found at:', candidate)
            return candidate

    # Try common mounted notebook roots like Colab and drive mounts
    roots = [Path('/content'), Path('/content/drive'), Path('/mnt'), Path.home()]
    for root in roots:
        if root.exists():
            candidate = root / 'data' / 'raw' / 'content_refresh_anonymized.csv'
            checked.append(candidate)
            if candidate.exists():
                print('DEBUG found at:', candidate)
                return candidate

    # Broad fallback search under likely roots
    for root in roots:
        if not root.exists():
            continue
        for candidate in root.rglob('content_refresh_anonymized.csv'):
            checked.append(candidate)
            if candidate.is_file():
                print('DEBUG found by rglob at:', candidate)
                return candidate

    raise FileNotFoundError(
        'Starter CSV not found. Checked: ' +
        ', '.join(str(p) for p in checked) +
        '; also searched: /content, /content/drive, /mnt, home dirs'
    )

path = starter_csv_path()
df = pd.read_csv(path)

# Quick sanity check for the starter dataset
print('rows:', len(df))
print('columns:', len(df.columns))
print('declining rows:', int((df['trend_direction'] == 'down').sum()))
print('median 90d impressions:', round(df['impressions_90d'].median(), 2))
print('mean CTR:', round(df['ctr'].mean(), 3))
```

## 2. The question: decision, action, cost of a wrong call

The decision I want to improve is which existing content pages should be prioritized for refresh. The action is a human reviewer or content team deciding where to spend limited time and effort on updates. A wrong recommendation could waste time on a page that is not a good candidate, or miss a page that could have benefited from refresh, which means slower gains and less efficient use of editorial effort. The problem is suitable for data and ML because the dataset contains measured signals that may indicate visibility decline, low engagement, or position-related opportunity before anyone manually selects a page.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

# Supporting numbers for the framing

df = pd.read_csv(starter_csv_path())

high_visibility = df['impressions_90d'] >= 100
declining_and_visible = (df['trend_direction'] == 'down') & high_visibility

print('Pages with at least 100 90d impressions:', int(high_visibility.sum()))
print('Declining pages among those:', int(declining_and_visible.sum()))
print('Share of visible pages that are declining:', round(declining_and_visible.mean() * 100, 1), '%')
```

## 3. Quick look at the data (2-3 real numbers)

A few starter-dataset numbers make this lane look worth exploring:

- The dataset has 30,000 rows, so there is enough coverage to look for stable patterns rather than relying on a handful of examples.
- About 54.2% of rows are labeled as declining, which suggests the problem is not rare and there is meaningful variation to study.
- Among pages with at least 100 impressions in the last 90 days, 59.8% are declining, which makes the refresh-priority question relevant for content with real visibility.

These numbers do not prove the final model will work, but they do suggest that the lane has enough signal and enough examples to justify the next 7 weeks of work.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

# Load the starter dataset and compute the supporting numbers used above

df = pd.read_csv(starter_csv_path())

rows = len(df)
declining_share = (df['trend_direction'] == 'down').mean() * 100
visible_pages = df['impressions_90d'] >= 100
declining_visible_share = ((df['trend_direction'] == 'down') & visible_pages).mean() * 100

print('rows:', rows)
print('declining share:', round(declining_share, 1), '%')
print('declining share among visible pages:', round(declining_visible_share, 1), '%')
```
